```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def l1(matrix: np.ndarray) -> np.float64:
    return np.max(np.sum(np.abs(matrix), axis=0))

def l2(matrix: np.ndarray) -> np.float64:
    return np.sqrt(np.max(np.linalg.eigvals(matrix.T @ matrix)))

def linf(matrix: np.ndarray) -> np.float64:
    return np.max(np.sum(np.abs(matrix), axis=1))

def frob(matrix: np.ndarray) -> np.float64:
    return np.sqrt(np.sum(matrix ** 2, axis=(0, 1)))
```

```python
def random_matrix(n: int) -> np.ndarray:
    return np.random.rand(n, n)


N = [10, 20, 50, 100, 200, 500]
R1, R2, R3, R4, R5, R6 = [], [], [], [], [], []
ratios_labels = [
    r'$R_1 = \frac{\|A\|_F}{\|A\|_2}$',
    r'$R_2 = \frac{\|A\|_1}{\|A\|_2}$',
    r'$R_3 = \frac{\|A\|_\infty}{\|A\|_2}$',
    r'$R_4 = \frac{\|A\|_1}{\|A\|_F}$',
    r'$R_5 = \frac{\|A\|_\infty}{\|A\|_F}$',
    r'$R_6 = \frac{\|A\|_1}{\|A\|_\infty}$'
]

for n in N:
    r1, r2, r3, r4, r5, r6 = [], [], [], [], [], []
    for _ in range(50):
        matrix = random_matrix(n)
        r1.append(frob(matrix) / l2(matrix))
        r2.append(l1(matrix) / l2(matrix))
        r3.append(linf(matrix) / l2(matrix))
        r4.append(l1(matrix) / frob(matrix))
        r5.append(linf(matrix) / frob(matrix))
        r6.append(l1(matrix) / linf(matrix))
    R1.append(np.round(np.mean(r1), 3))
    R2.append(np.round(np.mean(r2), 3))
    R3.append(np.round(np.mean(r3), 3))
    R4.append(np.round(np.mean(r4), 3))
    R5.append(np.round(np.mean(r5), 3))
    R6.append(np.round(np.mean(r6), 3))
```

```python
print("|   n   |   R1   |   R2   |   R3   |   R4   |   R5   |   R6   |")
for i in range(len(N)):
    print("|" + str(N[i]).center(7, " "), end="|")
    print(str(R1[i]).center(8, " "), end="|")
    print(str(R2[i]).center(8, " "), end="|")
    print(str(R3[i]).center(8, " "), end="|")
    print(str(R4[i]).center(8, " "), end="|")
    print(str(R5[i]).center(8, " "), end="|")
    print(str(R6[i]).center(8, " "), end="|")
    print()

all_ratios = [R1, R2, R3, R4, R5, R6]
alpha = []
print("|   a   |", end="")
for i in range(len(N)):
    current_R = eval(f"R{i + 1}")
    alpha.append(np.round(np.log(current_R[-1] / current_R[-2]) / np.log(N[-1] / N[-2]), 3))
    print(str(alpha[i]).center(8, " "), end="|")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, (ax, ratio, label) in enumerate(zip(axes, all_ratios, ratios_labels)):
    ax.loglog(N, ratio, 'bo-', linewidth=2, markersize=8)

    x_fit = np.array([N[0], N[-1]])
    y_fit = ratio[-1] * (x_fit / N[-1]) ** alpha[idx]
    ax.loglog(x_fit, y_fit, 'r--', linewidth=2,
              label=f'α = {alpha[idx]}')

    ax.set_xlabel('n')
    ax.set_ylim(0.8, 1.5)
    ax.set_ylabel('R')
    ax.set_title(label)
    ax.grid(True, which="both", ls="-", alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```
