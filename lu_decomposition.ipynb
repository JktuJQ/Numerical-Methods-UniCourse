```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def lu_decomposition(A):
    n = A.shape[0]
    L = np.zeros((n, n))
    U = np.zeros((n, n))

    for i in range(n):
        L[i, i] = 1.0
        for j in range(i, n):
            U[i, j] = A[i, j] - L[i, :i] @ U[:i, j]
        for j in range(i + 1, n):
            if abs(U[i, i]) < 1e-12:
                raise ValueError("LU-декомпозиция невозможна")
            L[j, i] = (A[j, i] - L[j, :i] @ U[:i, i]) / U[i, i]
    return L, U
```

```python
def relative_error(A, L, U):
    return np.linalg.norm(A - L @ U, 'fro') / np.linalg.norm(A, 'fro')


n = 10
alpha_values = np.logspace(-7, 2, 30)
errors = []
for alpha in alpha_values:
    A = np.random.randn(n, n)
    row_sums = np.sum(np.abs(A), axis=1) - np.abs(np.diag(A))
    np.fill_diagonal(A, alpha * row_sums)

    try:
        L, U = lu_decomposition(A)
        errors.append(relative_error(A, L, U))
    except ValueError:
        errors.append(np.nan)
```

```python
plt.figure(figsize=(10, 6))
plt.semilogx(alpha_values, errors, 'b-', linewidth=1.5)
plt.xlabel(r'$\alpha$', fontsize=12)
plt.ylabel('Относительная ошибка', fontsize=12)
plt.title('Зависимость относительной ошибки декомпозиции от параметра диагонального преобладания', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```
